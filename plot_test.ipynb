## 2. Interactive figure

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(-5, 25, 1000)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(t, np.exp(-((t - 5) / 6) ** 2) * np.cos(2 * np.pi * 0.5 * t))
ax.set_xlabel('t (ps)')
ax.set_ylabel('signal')
```

## 3. Widget controlling a plot

```python
import ipywidgets as widgets

fig2, ax2 = plt.subplots(figsize=(7, 3))
line, = ax2.plot(t, np.cos(2 * np.pi * 0.5 * t))
ax2.set_ylim(-1.1, 1.1)

def update(freq_THz=0.5):
    line.set_ydata(np.cos(2 * np.pi * freq_THz * t))
    fig2.canvas.draw_idle()

widgets.interact(update, freq_THz=(0.1, 2.0, 0.05));
```
